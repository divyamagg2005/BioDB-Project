# ClinVar Exploration

This notebook reproduces the Phase 1 ClinVar inspection report from the official raw archive. It does not perform final label filtering and does not modify the raw file.

```python
import json
import sys
from pathlib import Path

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(project_root / 'src'))
from load_clinvar import DEFAULT_INPUT, generate_report
```

```python
# This scans the gzip archive in a streaming pass and may take several minutes.
report = generate_report(project_root / DEFAULT_INPUT)
print(f"Rows: {report['row_count']:,}")
print(f"Columns: {report['column_count']}")
print(f"File size: {report['file_size_bytes']:,} bytes")
```

```python
print('Clinical significance distribution:')
display(report['clinical_significance_distribution'])
print('Variant type distribution:')
display(report['variant_type_distribution'])
print('Chromosome distribution:')
display(report['chromosome_distribution'])
print('Review status distribution:')
display(report['review_status_distribution'])
```

```python
missing_summary = {
    column: values for column, values in report['missing_values'].items()
    if values['count'] > 0
}
display(missing_summary)
```
